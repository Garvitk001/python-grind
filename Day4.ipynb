SORTING ALGORITHMS

Bubble sort

```python
# ---- BUBBLE SORT ----
# Logic: bar bar adjacent elements compare karo
# Bada element bubble ki tarah upar aata jaata hai
# Time complexity: O(n²)
```

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0,n-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1] , arr[j]
    return arr

numbers = [64, 34, 25, 12, 22, 11, 90]
print(f"Original  : {numbers}")
print(f"Sorted    : {bubble_sort(numbers.copy())}")
```

```python
marks = [85, 45, 92, 38, 78, 55, 91]
print(f"\nMarks Original : {marks}")
print(f"marks Sorted   : {bubble_sort(marks.copy())}")
```

Selection Sort

```python
# ---- SELECTION SORT ----
# Logic: har baar sabse chhota element dhundho
# usse sahi jagah rakho
# Time complexity: O(n²)
```

```python
def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i    
        
        for j in range(i+1, n):
            if arr[j] < arr[min_index]:
                min_index = j    
        arr[i], arr[min_index] = arr[min_index], arr[i]
    
    return arr


```

```python
original = [64, 34, 25, 12, 22, 11, 90]

print(f"Original         : {original}")
print(f"Bubble Sort      : {bubble_sort(original.copy())}")
print(f"Selection Sort   : {selection_sort(original.copy())}")
```

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]        
        j = i - 1
        
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]   
            j -= 1
        
        arr[j+1] = key      
    
    return arr
```

```python
print(f"Insertion Sort   : {insertion_sort(original.copy())}")
```

Insertion Sort  | Cards jaisa insert karo| Almost sorted data

# SEARCHING

```python
# ---- LINEAR SEARCH ----
# Ek ek element check karo
# Time complexity: O(n)


def linear_search(arr,target):
    for i , element in enumerate(arr):
        if element == target:
            return f"Target found in index : {i}"
    return f"Target not found in array"



# ---- BINARY SEARCH ----
# Sirf sorted array pe kaam karta hai
# Har baar array ko aadha karo
# Time complexity: O(log n)


def binary_search(arr,target):
    left = 0
    right= len(arr)-1

    while left <= right:
        mid = (left + right)//2

        if arr[mid] == target:
            return f" {target} mila — index {mid} pe"
        elif arr[mid] < target:
            left = mid + 1             
        else:
            right = mid - 1            
    
    return f" {target} nahi mila!"


numbers = [11, 12, 22, 25, 34, 64, 90]

print("--- Linear Search ---")
print(linear_search(numbers, 25))
print(linear_search(numbers, 100))

print("\n--- Binary Search ---")
print(binary_search(numbers, 25))
print(binary_search(numbers, 100))
```

# STACKS + QUEUES

```python
# ---- STACK ----
# LIFO = Last In First Out
# Real use: Undo/Redo, Browser back button, 
#           Function call stack
```

```python
class Stack:
    def __init__(self):
        self.items = []
    
    def push(self,item):
        self.items.append(item)
        return f"{item}  Pushed in stack"
    
    def pop(self):
        if self.is_empty():
            return f"Stack is empty"
        return f" {self.items.pop()} is pop from stack"
    
    def peek(self):
        if self.is_empty():
            return f"Stack is empty"
        return f"Top element {self.items[-1]}"
    
    def is_empty(self):
        return len(self.items) == 0
    
    def size(self):
        return len(self.items)
```

```python
browser = Stack()

print("--- Browser History ---")
print(browser.push("google.com"))
print(browser.push("github.com"))
print(browser.push("linkedin.com"))
print(browser.peek())             # current page
print(browser.pop())              # back button
print(browser.pop())              # back button
print(browser.peek())  
print(browser.pop())    
print(browser.pop())         
```

```python
# ---- QUEUE ----
# FIFO = First In First Out
# Real use: Print queue, Customer service,
#           Food delivery orders
```

```python
from collections import deque

class Queue:
    def __init__(self):
        self.items= deque()
    
    def enqueue(self,item):
        self.items.append(item)
        return f" {item} is enqueued in queue"
    
    def dequeue(self):
        if self.is_empty():
            return f" queue is empty"
        return f" {self.items.popleft()} is dequeue from queue"
    
    def front(self):
        if self.is_empty():
            return " Queue empty hai!"
        return f"Next: '{self.items[0]}'"
    
    def is_empty(self):
        return len(self.items) == 0
```

```python
orders = Queue()
print("\n--- Zomato Orders ---")
print(orders.enqueue("Garvit's order"))
print(orders.enqueue("Rahul's order"))
print(orders.enqueue("Priya's order"))
print(orders.front())              # next order kaun sa
print(orders.dequeue())            # pehla order deliver karo
print(orders.dequeue())            # doosra order
print(orders.front())   
```

```python
def check_balanced(expression):
    stack = []
    pairs = {')': '(', '}': '{', ']': '['} 
    
    for char in expression:
        if char in '({[':
            stack.append(char)        
        elif char in ')}]':
            if not stack:             
                return " Not Balanced"
            if stack[-1] != pairs[char]:  
                return " Not Balanced"
            stack.pop()               
    
    if stack:
        return " Not Balanced — opening brackets remaining"
    return " Balanced!"

# Test karo
expressions = ["(())", "({[]})", "(((", "(]", ")(", "({})"]
for exp in expressions:
    print(f"{exp:<10} → {check_balanced(exp)}")
```

# Student Merit List System using DSA

```python
students = [
    {"name": "Rahul Sharma",   "marks": 445, "rank": 0},
    {"name": "Garvit Khuteta", "marks": 498, "rank": 0},
    {"name": "Priya Singh",    "marks": 521, "rank": 0},
    {"name": "Amit Verma",     "marks": 389, "rank": 0},
    {"name": "Deepak Gupta",   "marks": 467, "rank": 0},
    {"name": "Neha Joshi",     "marks": 512, "rank": 0},
    {"name": "Rohan Meena",    "marks": 434, "rank": 0},
]
```

```python
def generate_merit_list(students):
    n = len(students)
    
    for i in range(n):
        for j in range(0, n-i-1):
            if students[j]["marks"] < students[j+1]["marks"]:  
               
                students[j], students[j+1] = students[j+1], students[j]

    for i, student in enumerate(students, 1):
        student["rank"] = i
    
    return students


sorted_students = generate_merit_list(students)
for s in sorted_students:
    print(f"Rank {s['rank']} | {s['name']:<18} | {s['marks']}")
```

```python
def search_student(students, target_marks):
    left  = 0
    right = len(students) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        if students[mid]["marks"] == target_marks:
            return f" Mila! {students[mid]['name']} — Rank {students[mid]['rank']} — Marks {students[mid]['marks']}"
        elif students[mid]["marks"] > target_marks:
            left = mid + 1     
        else:
            right = mid - 1
    
    return f" {target_marks} marks wala student nahi mila!"
```

```python

sorted_students = generate_merit_list(students)

print(search_student(sorted_students, 521))   
print(search_student(sorted_students, 498))   
print(search_student(sorted_students, 999))   
```

```python
from collections import deque

def get_waiting_list(students, selected=3):
    waiting = deque()
    
    for student in students[selected:]:
        waiting.append(student["name"])
    
    return waiting


waiting_list = get_waiting_list(sorted_students)
print(" Waiting List:")
while waiting_list:
    print(f"   -> {waiting_list.popleft()}")
```

```python
def print_merit_list(students, selected=3):
    print("=" * 50)
    print("           MERIT LIST 2024")
    print("=" * 50)
    print(f"{'Rank':<6} {'Name':<20} {'Marks':<8} Status")
    print("-" * 50)
    
    for student in students:
        if student["rank"] <= selected:
            status = " Selected"
        elif student["rank"] <= selected + 3:
            status = " Waiting"
        else:
            status = " Not Selected"
        
        print(f"{student['rank']:<6} {student['name']:<20} {student['marks']:<8} {status}")
    
    print("=" * 50)
    print(f" Topper: {students[0]['name']} ({students[0]['marks']} marks)")
    
    
    waiting_names = [s['name'].split()[0] for s in students[selected:selected+3]]
    print(f" Waiting : {', '.join(waiting_names)}")
    print("=" * 50)


print_merit_list(sorted_students)
```
