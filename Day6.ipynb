# Session -2 Pandas

```python
import pandas as pd

data = {
    "name"    : ["Garvit", "Rahul", "Priya", "Amit", "Deepak", "Neha"],
    "age"     : [22, 21, 22, 23, 21, 22],
    "city"    : ["Jaipur", "Delhi", "Mumbai", "Jaipur", "Delhi", "Mumbai"],
    "salary"  : [45000, 55000, 62000, 38000, 51000, 58000],
    "cgpa"    : [8.3, 7.5, 9.2, 6.8, 7.8, 8.9],
    "joining" : ["2024-01-15", "2024-02-20", "2024-01-10", 
                 "2024-03-05", "2024-02-15", "2024-01-25"]
}

df = pd.DataFrame(data)
print(f"=====Basic Info=====")
print(f"Shape  : {df.shape}")
print(f"Column : {list(df.columns)}")
print(f"\nFirst 3 rows:\n {df.head(3)}")
print(f"\nData types:")
print(df.dtypes)
print(f"\nBasic stats:")
print(df.describe())
```

 DATA SELECTION + FILTERING

```python
# Column Selection 

print(f"{df[["name","salary","city"]]}") 
```

```python
#Row filtering 

jaipur = df[df["city"] == "Jaipur"]
print(f"Japur Employees :\n {jaipur[["name","salary"]]}")
```

```python
high_salary = df[df["salary"] >=50000]
print(f"Salary greater than 50k :\n {high_salary[["name","salary"]]}")
```

```python
good_employees= df[(df["cgpa"] >=8)  & (df["salary"]>= 50000)]
print(f"Good Employees : \n {good_employees[["name","cgpa","salary"]]}")
```

```python
print("\n=== SORTING ===")
by_salary = df.sort_values("salary", ascending=False)
print(by_salary[["name", "salary"]].head(5))
```

Data Cleaning

```python
dirty_data = {
    "name"   : ["Garvit", "Rahul", None, "Amit", "Deepak", "Neha"],
    "age"    : [22, None, 22, 23, 21, 22],
    "salary" : [45000, 55000, 62000, None, 51000, 58000],
    "city"   : ["Jaipur", "Delhi", "Mumbai", "jaipur", "DELHI", "mumbai"],
}

df2 = pd.DataFrame(dirty_data)

print("=== DIRTY DATA ===")
print(df2)
print(f"\nNull values:\n{df2.isnull().sum()}")
print("\n=== CLEANING ===")

# Name null — row drop karo
df2 = df2.dropna(subset = ["name"])


df2["age"] = df2["age"].fillna(df2["age"].mean())
df2["salary"] = df2["salary"].fillna(df2["salary"].median())
df2["city"] = df2["city"].str.strip().str.title()

print(df2)

print(f"Null Value : \n {df2.isnull().sum()}")
```

GroupBy + Aggregation

```python
#group by           
df2_groupby =df2.groupby("city")["salary"].mean()
print(df2_groupby)

#aggregration
df2_agg =df2.groupby("city")["salary"].agg(["mean","max","min","std"])
print(df2_agg)
```
