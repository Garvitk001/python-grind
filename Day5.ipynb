```python
# ============================================
# ERROR HANDLING - try/except
# Real use: Production code mein crash rokna
# ============================================

# Bina error handling ke — crash ho jaata hai
# print(int("Garvit"))  # ValueError — program band!
```

```python
def safe_divide(a,b):
    try:
        result = a/b
        return f"Result : {result}"
    except ZeroDivisionError:
        return "Zero se divide nahi kar sakte!"
    except TypeError:
        return "Sirf no. allowed h"
    finally:
        print("(callculation attempted)")
    
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "a"))
```

# Safe Student Data Loader:

```python
def load_student_data(filename):
    try:
        with open(filename,"r") as file:
            for line in file:
                try:
                    parts=line.strip().split(",")
                    name =parts[0]
                    cgpa = float(parts[2])
                    if 0<= cgpa <=10:
                        print(f"{name} loaded-Cgpa: {cgpa}")
                    else:
                        print(f"{name} skip - Invalid CGPA")
                except ValueError:
                   print("Line skip invalid data")
    except FileNotFoundError:
        print("File Not found")
    finally:
        print("file processing complete")

load_student_data("students.txt")
```

# Session 2 — RECURSION

Recursion -Function khud apne aap ko call kare — jab tak answer na mil jaaye!

```python
# ---- BASIC RECURSION ----
# 2 cheezein zaroori hain:
# 1. Base case  — kab rukna hai
# 2. Recursive case — khud ko call karo
```

```python
#factorial

def factorial(n):
    if n==0 or n==1:              #base case
        return 1
    
    return n * factorial(n-1)        #recurrisive case

# factorial(n-1) = factorial(4)  and this call itaself this function

print(factorial(5))
```

```python
# Fibonacci

def fibonacci(n):
    if n <=1:
        return n 
    return fibonacci(n-1) + fibonacci(n-2)
print("====Fibonacci Series====")
for i in range(10):
    print(fibonacci(i),end=" ")
```

```python
#sum of digit

def sum_of_digit(n):
    if n <10:
        return n
    return n%10 + sum_of_digit(n//10)

print(f"\n\nSum of digits of 1234 : {sum_of_digit(1234)}")  #
print(f"Sum of digits of 9876 : {sum_of_digit(9876)}")
```

```python
# Powe Calculation

def power(base,exp):
    if exp==0:
        return 1
    return base * power(base,exp-1)

print(f"3^10: {power(3,10)}")
```

```python
#countdown from reverse

def count_down(n):
    if n==0:
        print (f"0 Blast off")
        return
    print (n,end=" ")
    count_down(n-1)
(count_down(10))
```

```python
# max element in list
def find_max(arr):
    if len(arr) == 1:        
        return arr[0]
    
    rest_max = find_max(arr[1:])   
    
    if arr[0] > rest_max:
        return arr[0]
    else:
        return rest_max

print(find_max([3, 7, 1, 9, 4])) 
```
