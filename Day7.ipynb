```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
print(f"===Titanic Dataset===")
print(f"Shape : {df.shape}")
print(f"Clolumns : {list(df.columns)}")
print(df.head(7))
```

```python
print(df.isnull().sum())
print(df.describe)
```

```python
survival_rate = df["Survived"].mean() * 100
print(f"Survival Rate: {survival_rate:.2f}%")
```

```python
gender_survival = df.groupby("Sex")["Survived"].mean()*100
print(gender_survival)
```

```python
plt.figure(figsize=(6,5))
plt.bar(gender_survival.index, gender_survival.values, 
        color=["coral","steelblue"], edgecolor="black")
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate %")
plt.tight_layout()
plt.show()
```

```python

Pclass_survial= df.groupby("Pclass")["Survived"].mean()*100
plt.figure(figsize=(8,6))
plt.bar(Pclass_survial.index, Pclass_survial.values,color=["gold","steelblue","coral"], edgecolor="black")
plt.title("Survival Rate by Class")
```

```python

df["Age"] = df["Age"].fillna(df["Age"].mean())

df = df.drop(columns=["Cabin"])


df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])


print("Null values after cleaning:")
print(df.isnull().sum())
print(f"\nShape after cleaning: {df.shape}")
```
